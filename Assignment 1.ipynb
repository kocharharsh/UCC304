```python
#6.1
def prime(n):
  sum =0
  for i in range(1,n+1):
    flag=0
    for j in range(2,int(i**1/2 +1)):
      if i%j==0:
        flag=1
        break

    if flag==0:
      sum+=i
  return sum

print (prime(10))
```

```python
# 6.2
def odd_sum(n):
  sum =0
  for i in range(1,n+1):
    if (i%2!=0):
      sum+=i
  return sum

print (odd_sum(10))
```
