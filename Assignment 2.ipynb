```python
import random as r
r.seed(32)
rnd_list=[r.randint(100,900)for i in range (100)]
print(rnd_list)

odd=[x for x in rnd_list if x%2!=0]
print(odd)
print("There are",len(odd),"odd numbers random list")

even=[x for x in rnd_list if x%2==0]
print(even)
print("There are",len(even),"even numbers random list")

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_list = [num for num in rnd_list if is_prime(num)]
print(prime_list)
frequency = {}

for num in rnd_list:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

ans = rnd_list[0]
max = frequency[ans]

for num in frequency:
    if frequency[num] > max:
        ans = num
        max = frequency[num]

print("The most frequent number is", ans)
print("It occurs", max, "times")
```

```python
roll = "1024170032"

digits = [int(digit) for digit in roll]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A ", A)
print("Set B ", B)

union = A.union(B)

print("Union of A and B ", union)

intersection = A.intersection(B)

print("Intersection of A and B ", intersection)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B ", A_minus_B)
print("B - A ", B_minus_A)


symmetric = A.symmetric_difference(B)

print("Symmetric difference:", symmetric)

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

x = int(input("Enter a value to remove from A: "))
A.discard(x)

print("After discaarding ",A)
```

```python
my_dict = {
    "name": "Harsh Kochar",
    "roll_no": "1024170032",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}
print(my_dict)

my_dict["location"] = my_dict.pop("city")

my_dict["cgpa"] = 9.39

my_dict["age"] = my_dict["age"] + 1

print(my_dict)

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("\nDictionary after using pop():")
print(dict_pop)

del dict_del["branch"]

print("\nDictionary after using del:")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail is not present in the dictionary")


friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024170035",
    "branch": "CSE",
    "age": 19,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

string_dict = {key: value for key, value in my_dict.items()
               if isinstance(value, str)}

print("\nDictionary containing only string values:")
print(string_dict)
```
