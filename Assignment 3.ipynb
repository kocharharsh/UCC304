```python
def cat(r):
  if r['Rating']>=4.5:
    return "Excelent"
  elif r['Rating']>=4:
    return "Good"
  else :
    return "Average"
df_emp["Category"]=df_emp.apply(cat,axis=1)
df_emp
```

```python
# h) Identify missing values in the DataFrame.
print(df_emp.isnull().sum())
```

```python
# i) Rename the Employee_ID column to ID.
df_emp=df_emp.rename(columns={'Employee_ID':'Id'})
df_emp
```

```python
# j) Find all employees who:
# i. Have more than 5 years of experience.
df_exp=df_emp[df_emp['Years_of_Experience']>5]
df_exp
```

```python
# ii. Belong to the IT department.
df_it=df_emp[df_emp['Department']=='IT']
df_it
```

```python
# k) Modify the dataset by adding a new column, Tax, which deducts 10% of the
# Salary.
df_emp['Tax']=df_emp['Salary']*0.10
df_emp
```
