## Q1. Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like: 

### a) Addition of 2 in all the elements

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

result = arr + 2
print(result)
```

### b) Multiply 3 with all the elements

```python
result = arr * 3
print(result)
```

### c) Divide every element by 2

```python
result = arr / 2
print(result)
```

## Q2. Basic NumPy Array

### a) Reverse the NumPy array

```python
arr = np.array([1, 2, 3, 6, 4, 5])

reversed_arr = arr[::-1]
print(reversed_arr)
```

### b) Find the most frequent value and its indices

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

## Q3. Access Elements from a 2-D Array

## Q4. NumPy Array Using `linspace()`

```python
Harsh = np.linspace(10, 100, 25)

print("Array:", Harsh)
print("Dimensions:", Harsh.ndim)
print("Shape:", Harsh.shape)
print("Total elements:", Harsh.size)
print("Data type:", Harsh.dtype)
print("Total bytes:", Harsh.nbytes)

transpose_Harsh = Harsh.reshape(25, 1)
print("\nTranspose using reshape():")
print(transpose_Harsh)

print("\nTranspose using T attribute:")
print(Harsh.T)

print("\nCan T be used directly?")
print("For a 1-D NumPy array, .T does not change its shape.")
print("It returns the same 1-D array. Use reshape(25, 1) for a column form.")
```

## Q5. NumPy 2-D Array Operations

```python
ucs420_Harsh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Array:")
print(ucs420_Harsh)

print("\nMean:", np.mean(ucs420_Harsh))
print("Median:", np.median(ucs420_Harsh))
print("Maximum:", np.max(ucs420_Harsh))
print("Minimum:", np.min(ucs420_Harsh))
print("Unique elements:", np.unique(ucs420_Harsh))
```

```python
reshaped_ucs420_Harsh = ucs420_Harsh.reshape(4, 3)

print("Reshaped array (4 x 3):")
print(reshaped_ucs420_Harsh)
```

```python
resized_ucs420_Harsh = np.resize(ucs420_Harsh, (2, 3))

print("Resized array (2 x 3):")
print(resized_ucs420_Harsh)
```
