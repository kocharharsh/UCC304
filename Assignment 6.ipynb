## Q1. Sensor Readings

Consider the following sensor readings:

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```

A sensor has an average measurement error of **+2°C**.

Perform the following without using a `for` loop:

**a.** Add 2°C to every reading using vectorization.  
**b.** Convert all temperatures from Celsius to Fahrenheit using:

\[
F = rac{9}{5}C + 32
\]

**c.** Identify all readings greater than 32°C using Boolean indexing.  
**d.** Count how many readings exceed 32°C.  
**e.** Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.

````python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C to every reading
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

# b) Convert Celsius to Fahrenheit
fahrenheit = (9 / 5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

# c) Readings greater than 32°C
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

# d) Count readings exceeding 32°C
count = np.sum(corrected_temperature > 32)
print("Number of readings exceeding 32°C:", count)
````

**e) Explanation:** NumPy vectorization applies an operation to the entire array at once, avoiding an explicit `for` loop. Boolean indexing creates a Boolean mask and directly selects elements satisfying a condition. These operations are concise and are generally more efficient for numerical array processing.

## Q2. Daily Steps

The following matrix represents the number of daily steps recorded by four users over three days:

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])
```

**a.** Calculate the total steps recorded.  
**b.** Calculate the mean number of steps.  
**c.** Find the maximum and minimum values.  
**d.** Calculate the total steps for each day using `axis=0`.  
**e.** Calculate the total steps for each user using `axis=1`.  
**f.** Find the user/day position containing the maximum number of steps using `argmax()`.

````python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps
print("Total steps:", np.sum(steps))

# b) Mean steps
print("Mean steps:", np.mean(steps))

# c) Maximum and minimum
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# d) Total steps for each day
print("Total steps for each day:", np.sum(steps, axis=0))

# e) Total steps for each user
print("Total steps for each user:", np.sum(steps, axis=1))

# f) Position of maximum value
max_index = np.argmax(steps)
position = np.unravel_index(max_index, steps.shape)
print("User/day position of maximum steps:", position)
print("Maximum value:", steps[position])
````

original = np.array([1, 2, 3, 4, 5, 6])

## Q4. Cognitive-Assistive System

A simple cognitive-assistive system wants to predict a user's assistance score based on three factors:

- Sleep hours
- Activity level
- Stress level

Consider:

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

Each row represents one user and each column represents a feature.

**a.** Display the shape and dimensions of `X`.  
**b.** Calculate `X.T` and explain what the transpose represents.  
**c.** Calculate the matrix product `X.T @ X`.  
**d.** Calculate the matrix inverse using `np.linalg.inv()` where appropriate.  
**e.** Implement the Ordinary Least Squares equation:

\[
eta = (X^T X)^{-1}X^T y
\]

**f.** Explain what the resulting coefficients represent in the context of the three user features.  
**g.** A new user has:

```python
new_user = np.array([5, 40, 7])
```

Use the calculated coefficients to obtain a predicted assistance score.

````python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a) Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b) Transpose
print("\nX.T:")
print(X.T)

# c) X.T @ X
XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

# d) Inverse
XtX_inv = np.linalg.inv(XtX)
print("\nInverse of X.T @ X:")
print(XtX_inv)

# e) Ordinary Least Squares
beta = XtX_inv @ X.T @ y
print("\nOLS coefficients:")
print(beta)

# g) Prediction for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\nPredicted assistance score:", predicted_score)
````

**b) Explanation:** `X.T` changes the matrix from shape `(4, 3)` to `(3, 4)`. The rows of the transposed matrix correspond to the three features: sleep hours, activity level, and stress level.

**f) Explanation:** The three coefficients correspond respectively to **sleep hours, activity level, and stress level**. Each coefficient indicates the change in the predicted assistance score associated with a one-unit change in that feature, while the other features are held constant.
